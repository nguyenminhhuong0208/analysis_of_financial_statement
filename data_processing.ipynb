```python
from docx import Document
import pandas as pd
```

```python
# Function to extract and preprocess data from tables
def process_tables_to_text(doc_path):
    doc = Document(doc_path)
    texts = []

    # Extract paragraphs
    for paragraph in doc.paragraphs:
        if paragraph.text.strip():
            texts.append(paragraph.text.strip())

    # Extract data from tables and convert to text
    for table in doc.tables:
        for row in table.rows:
            row_text = " | ".join([cell.text.strip() for cell in row.cells if cell.text.strip()])
            if row_text:
                texts.append(f"Table: {row_text}")  # Add prefix to identify table data

    return texts

# File path
doc_path = "Data/00002469135802469136ctcp-bcg-land03022025-000000bo-co-ti-chnh-qu-4-2024-ca-cng-ty-m.docx"

# Extract and convert data
processed_texts = process_tables_to_text(doc_path)
```

```python
# 2. Preprocess data
def preprocess_data(df):
    # Remove empty rows/columns
    df.dropna(how='all', inplace=True)
    df.dropna(axis=1, how='all', inplace=True)

    # Normalize text
    for col in df.columns:
        if df[col].dtype == 'object':  # Check the data type of each column
            df[col] = df[col].str.lower()  # Convert to lowercase
            df[col] = df[col].str.replace(r'[^\w\s]', '', regex=True)  # Remove special characters
            df[col] = df[col].str.strip()  # Remove extra spaces

    return df
```

```python
# 1. Extract all tables from a Word file
def extract_tables_from_doc(doc_path):
    doc = Document(doc_path)
    tables = doc.tables
    dataframes = []

    if tables:
        print(f"Total number of extracted tables: {len(tables)}")

        for i, table in enumerate(tables, start=1):
            # print(f"\nConverting Table {i} to DataFrame...")

            table_data = []
            for row in table.rows:
                row_data = [cell.text.strip() for cell in row.cells]
                table_data.append(row_data)

            df = pd.DataFrame(table_data[1:], columns=table_data[0])
            dataframes.append(df)

            # print(f"DataFrame of Table {i}:")
            # print(df)
    else:
        print("No tables found in the file.")

    return dataframes
```

```python
# 4. File path for the Word document
doc_path = "/content/00002469135802469136ctcp-bcg-land03022025-000000bo-co-ti-chnh-qu-4-2024-ca-cng-ty-m.docx"

# Extract and convert tables to DataFrames
dataframes = extract_tables_from_doc(doc_path)
```

```python
import pandas as pd
from transformers import BertTokenizer, BertForSequenceClassification
import torch

# Tải FinBERT mà không dùng context đặc biệt
tokenizer = BertTokenizer.from_pretrained('yiyanghkust/finbert-tone')
model = BertForSequenceClassification.from_pretrained('yiyanghkust/finbert-tone')
model.eval()  # Đặt mô hình ở chế độ đánh giá
```

```python
from transformers import pipeline
# Tạo pipeline cho text-classification
nlp = pipeline("text-classification", model=model, tokenizer=tokenizer)
```

```python
# Hàm chuyển DataFrame thành văn bản có ý nghĩa
def dataframe_to_text(df, statement_type):
    text = f"{statement_type} Analysis:\n"
    for index, row in df.iterrows():
        # Chuyển dòng thành câu văn
        row_text = " ".join([f"{col}: {val}" for col, val in row.items() if pd.notna(val)])
        text += f"{row_text}\n"
    return text

# Hàm phân tích cảm xúc bằng FinBERT
def analyze_sentiment(text):
    # Tokenize văn bản
    inputs = tokenizer(text, return_tensors="pt", truncation=True, max_length=512, padding=True)

    # Dự đoán với FinBERT
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits
        probs = torch.softmax(logits, dim=1)
        sentiment_scores = probs[0].tolist()  # [negative, neutral, positive]

    labels = ['Negative', 'Neutral', 'Positive']
    sentiment = labels[sentiment_scores.index(max(sentiment_scores))]
    return sentiment, sentiment_scores


# Xử lý và phân tích từng bảng
for df, statement_type in dataframes:
    # Chuyển DataFrame thành văn bản
    text = dataframe_to_text(df, statement_type)
    print(f"\n{statement_type} Text:")
    print(text)

    # Phân tích cảm xúc
    sentiment, scores = analyze_sentiment(text)
    print(f"\n{statement_type} Sentiment Analysis:")
    print(f"Sentiment: {sentiment}")
    print(f"Scores: Negative: {scores[0]:.4f}, Neutral: {scores[1]:.4f}, Positive: {scores[2]:.4f}")

# Phân tích tổng quan
combined_text = "\n".join([dataframe_to_text(df, statement_type) for df, statement_type in dataframes])
overall_sentiment, overall_scores = analyze_sentiment(combined_text)
print("\nOverall Financial Statement Sentiment Analysis:")
print(f"Sentiment: {overall_sentiment}")
print(f"Scores: Negative: {overall_scores[0]:.4f}, Neutral: {overall_scores[1]:.4f}, Positive: {overall_scores[2]:.4f}")
```

```python
from transformers import AutoTokenizer, AutoModel
model_name = "vinai/phobert-large"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
```

```python
# Hàm chuyển DataFrame thành văn bản có ý nghĩa
def dataframe_to_text_v2(df, statement_type):
    text = f"{statement_type} Analysis:\n"
    for index, row in df.iterrows():
        values = {}
        for col, val in row.items():
            if pd.notna(val) and not isinstance(val, str) and col not in ['Mã số', 'TÀI SẢN', 'Thuyết _ minh']:
                try:
                    float(val)  # Kiểm tra xem có phải là số không
                    values[col] = val
                except ValueError:
                    pass  # Bỏ qua nếu không phải là số

        if values:
            value_pairs = [f"{col}: {val}" for col, val in values.items()]
            text += f"Tại dòng {index}, các giá trị đáng chú ý là: {', '.join(value_pairs)}.\n"
    return text

# Chuyển toàn bộ list dataframes1 sang text
combined_text = ""
for df, statement_type in dataframes1:
    combined_text += dataframe_to_text(df, statement_type)
    combined_text += "\n" # Thêm một dòng trắng giữa các bảng cho dễ đọc
```

```python
def get_phobert_features(text):
    """Trích xuất đặc trưng từ văn bản bằng mô hình PhoBERT."""
    inputs = tokenizer(text, return_tensors="pt", max_length=512, truncation=True)

    with torch.no_grad():
        outputs = model(**inputs)

    # 'last_hidden_state' chứa các embedding của từng token
    last_hidden_state = outputs.last_hidden_state

    # Embedding của token [CLS] (biểu diễn cho toàn bộSequence)
    cls_embedding = last_hidden_state[:, 0, :]

    return last_hidden_state, cls_embedding

# Đưa combined_text vào PhoBERT để trích xuất đặc trưng
all_token_embeddings, cls_embedding = get_phobert_features(combined_text)

print("Nội dung của combined_text trước khi đưa vào PhoBERT:\n", combined_text)
all_token_embeddings, cls_embedding = get_phobert_features(combined_text)
```
